## GLOBAL TABLES

```python
MONTHLY_PATH = "/Workspace/bi_subject/economics_project/assignment1/data/final/economics_monthly.csv"
QUARTERLY_PATH = "/Workspace/bi_subject/economics_project/assignment1/data/final/economics_quarterly.csv"

monthly = spark.read.option("header", True).option("inferSchema", True).csv(MONTHLY_PATH)
quarterly = spark.read.option("header", True).option("inferSchema", True).csv(QUARTERLY_PATH)

monthly.write.format("delta").mode("overwrite").saveAsTable("bi_subject.colombia_economics.monthly")
quarterly.write.format("delta").mode("overwrite").saveAsTable("bi_subject.colombia_economics.quarterly")
```
